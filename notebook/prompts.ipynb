```python
from IPython.display import Markdown, display

def show_response(response):
    content = response.content

    if isinstance(content, str):
        text = content
    elif isinstance(content, list):
        text = "\n".join(
            block["text"]
            for block in content
            if isinstance(block, dict) and block.get("type") == "text"
        )
    else:
        text = str(content)

    display(Markdown(text))
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI
llm=ChatGoogleGenerativeAI(model='gemini-3.5-flash-lite')
```

```python
prompts=[
    {"role":'system','content':'You are a python developer.'},
    {"role":'user','content':'Difference between list and tuple? give me answer in the form of table and also give me output in 10 lines'}
]
res=llm.invoke(prompts)
show_response(res)
```

```python
prompts=[
    {"role":'system','content':'You are translator and translate input in hindi'},
    {"role":'user','content':'Hey!! I loved her more than myself..'}
]
res=llm.invoke(prompts)
show_response(res)
```

```python
prompts=[
    {"role":'system','content':'You are translator and translate input in german and also give me pronounciation of the translated text in english'},
    {"role":'user','content':'Hey!! I loved her more than myself..'}
]
res=llm.invoke(prompts)
show_response(res)
```

Prompt Templates - Dynamic Prompts

```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
```

```python
llm=ChatGoogleGenerativeAI(model='gemini-3.5-flash-lite')
```

```python
prompts=ChatPromptTemplate.from_messages([
    {'role':'system','content':'You are translator and translate input in {language}'},
    {'role':'user','content':'{query}'}
])
```

```python
#final_prompts=prompts.format_messages(language='french',query='Hey!! I loved her more than myself..')
#response=llm.invoke(final_prompts)
#show_response(response)
```

```python
prompts.invoke({'language':'french','query':'Hey!! I loved her more than myself..'})
result = llm.invoke(prompts.invoke({'language':'french','query':'Hey!! I loved her more than myself..'}))
show_response(result)
```

```python
def transform_output(result:str):
    return result.upper()
```

```python
str=StrOutputParser()
```

```python
chains = prompts | llm | str | transform_output
```

```python
result = chains.invoke({'language':'hinglish','query':'Hey!! I loved her more than myself..'})
print(result)
```
