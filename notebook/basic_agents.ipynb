```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_core.tools import tool
```

```python
@tool
def add_numbers(a:int,b:int):
    """
    it will return the sum of two numbers
    Args:
        a - number one
        b - number two
    """
    return a + b
```

```python
@tool
def multiply_numbers(a:int,b:int):
    """
    it will return the multiplication of two numbers
    Args:
        a - number one
        b - number two
    """
    return a * b
```

```python
from langchain.agents import create_agent
from langchain_groq import ChatGroq
llm = ChatGroq(model = 'openai/gpt-oss-20b')
```

```python
agent = create_agent(
    model = llm,
    tools=[add_numbers,multiply_numbers],
    system_prompt="You are a maths teacher and always use tools for calculation"
)
```

```python
response = agent.invoke({'messages':[{'role':'user','content':'what is 2+3*20?'}]})
print(response['messages'][-1].content)
```
